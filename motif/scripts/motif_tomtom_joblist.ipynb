```python
import os
import pandas as pd
import numpy as np
```

```python
sp = "fly"
for suffix in ["AllPeaks", "Top20Percent", "Top20PercentRandom_Rep1", "Top20PercentRandom_Rep2", "Top20PercentRandom_Rep3", "RemoveHOT", "RemoveHOTRandom_Rep1", "RemoveHOTRandom_Rep2", "RemoveHOTRandom_Rep3", "RemoveUltraHOT", "RemoveUltraHOTRandom_Rep1", "RemoveUltraHOTRandom_Rep2", "RemoveUltraHOTRandom_Rep3"]:

    files = pd.DataFrame(os.listdir("../result/motif/%sSTREME_%s/" % (sp.capitalize(), suffix)))
    files.loc[:, [1,2]] = files[0].str.split("_", expand=True, n=1).values
    motifs = [ii.replace(".meme", "") for ii in os.listdir("/home/jg2447/slayman/motifs/pwm/cisbp_InferAndDirect/%s_stack/" % sp)]
    files[3] = files[1].isin(motifs)

    valid = files.loc[files[3], 0].values
    valid_m = files.loc[files[3], 1].values

    # joblist of tomtom, comparing inferred motifs to real motifs
    os.makedirs("../result/tomtom/%sSTREME_%s/" % (sp.capitalize(), suffix), exist_ok=True)
    with open("./job_tomtom_%s_%s" % (sp, suffix), "w") as f:
        for ff, mm in zip(valid, valid_m):
            f.write("module load MEME; tomtom -oc ../result/tomtom/%sSTREME_%s/%s/ ../result/motif/%sSTREME_%s/%s/streme.txt /home/jg2447/slayman/motifs/pwm/cisbp_InferAndDirect/%s_stack/%s.meme\n" % (sp.capitalize(), suffix, ff, sp.capitalize(), suffix, ff, sp, mm))
```
